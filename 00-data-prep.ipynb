```python
from huggingface_hub import hf_hub_download

products_path = hf_hub_download(REPO, "digikala-products.csv",
                                repo_type="dataset", revision=REV)
comments_path = hf_hub_download(REPO, "digikala-comments.csv",
                                repo_type="dataset", revision=REV)
print(products_path, comments_path)
```

```python
import pandas as pd

products = pd.read_csv(products_path)
comments = pd.read_csv(comments_path)
print(products.shape, comments.shape)

products.to_parquet("/kaggle/working/products_raw.parquet", index=False)
comments.to_parquet("/kaggle/working/comments_raw.parquet", index=False)
```
